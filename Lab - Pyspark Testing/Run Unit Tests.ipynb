#### Para executar o pytest dentro do Databricks:

Por padrão, o Python gera arquivos de bytecode .pyc em um diretório pycache. 

Quando executa o pytest, ele procura por:

•	Arquivos de teste: Por padrão, arquivos chamados test_*.py ou *_test.py. 

•	Funções de teste: Funções começando com test_*(). 

Por exemplo, se você tiver um arquivo como test_example.py com uma função test_addition(), o pytest irá automaticamente detectar e executar os testes naquele arquivo.

```python
import pytest
import sys
```

```python
sys.dont_write_bytecode = True 
#Definir isso como True previne isso, o que é útil em ambientes somente leitura. 

retcode = pytest.main(
    [
    'Unit_Tests/test_Spark_Helper_Functions.py',
    '-v',               # -v = verbose 
                        #fornece uma saída mais detalhada durante a execução dos testes, incluindo os nomes dos testes e seus resultados. 
    '-p',   
    'no:cacheprovider'  # desabilita o provedor de cache do pytest.
    ]
)

#Resultado do teste 
# 0     → sucesso 
# <> 0  → falha 
assert retcode == 0
```
